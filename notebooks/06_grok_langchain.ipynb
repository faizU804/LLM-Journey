```python
from langchain_groq import ChatGroq
from langchain_openai import ChatOpenAI
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b")
openaillm = ChatOpenAI(model="gpt-5-mini-2025-08-07")
```

```python
res = llm.invoke("who is pm of pakistan")
print(res.content)
```
