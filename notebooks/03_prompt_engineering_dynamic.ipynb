Prompts templates we need bcz each time code repeated its not well for production level

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatOpenAI(model="gpt-5-mini-2025-08-07")
out = StrOutputParser()
```

chain node we make here 

```python
def UpperCaseMaker(result:str):
    return result.upper()
```

```python
prompts = ChatPromptTemplate.from_messages([
     { "role" : "system" , "content":"You are translator and translate input {language}"},
        {"role":"user", "content": "{query} "}
])
final_prompts = prompts.format_messages(language = "urdu" , query = "I love python and javascripts")
# final_prompts 
```

know we pass this final prompt to the llm to invoke the function and get the result 

```python
res = llm.invoke(final_prompts)
print(res.content)
```

why we write llm inovke and other above code for this to get llm result . know we go through with standard approch and thats why we use chaining


```python
res = llm.invoke(prompts.invoke({"language" : "Urdu" , "query": "I love Machine learning "}))
res.content
```

correct way to creat chain just like production level

```python
chains = prompts | llm | out | UpperCaseMaker
```

```python
chains
```

```python
res = chains.invoke({"language" : "Hinglish" , "query": "I love Machine learning "})
# print(res.content) after using out we dont need content again and again to get the value 
print(res)
```
