```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = [
    "Machine learning is a subset of artifical intelligence",
    "Deep learning uses neural networks with multiple layers",
    "python is a progrmming language popular for database system",
    "Kashmir never compromise on their soverity its never done",
    "AI models can understand and generate human language"
]
```

```python
from langchain_openai import OpenAIEmbeddings
```

```python
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
# embeddings = OpenAIEmbeddings(model="text-embedding-3-large")
```

```python
vector = embeddings.embed_documents(documents)
len(vector)
# len(vector[0])
```

store vector embeddings to vector db phase is here .

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(
    texts=documents,
    embedding=embeddings,
    persist_directory="./vector_db"
)
```

Next important step is base on user query how we give answer related to their query . here we do this after saving embeddings vectors into db.

```python
# query = "Machine Learnings"
query = "Data science"
# result = vector_store.similarity_search(query=query)
result = vector_store.similarity_search(query=query , k=2)
result
```
