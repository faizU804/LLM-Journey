```python
prompts = [
    {"role":"user" , "content":"Hello , My name is Faiz ur Rehman"},
    {"role":"ai" , "content":"Hello Faiz ur Rehman — nice to meet you! How can I help you today?"},
    {"role" : "user" , "content":"What is my name ?"}
    ]

res = llm.invoke(prompts)
res.content
```

Production way to manage the history of chat so they remember them and use them as well previous memory

```python
history = []   #first store just like list of dictionary [{role:user , content : what is hosuehold}]

while True:
    query = input("User : ")
    if query.lower() in ["exit" , "byee" , "quit"]:
        print("Good by see you later")
        break
    
    
    history.append({"role" : "user" , "content":query})
    print("User : " , query)
    
    
    res = llm.invoke(history)
    history.append({"role" : "ai" , "content":res.content})
    print("AI : "  , res.content, "\n")
else:
    print("Chatting is End know ! ")
```
