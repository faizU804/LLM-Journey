```python
from langchain_core.tools import tool
```

Here we make tool that we use for over work 

```python
@tool
def add_num(a:int , b:int):
    """
        its will return the sum of the two numbers
        
        Args: 
            a : Number One
            b : Number Two
    """
    return a + b 


@tool
def multiply_num(a:int , b:int):
    """
        its will return the multiply of the two numbers
        
        Args: 
            a : Number One
            b : Number Two
    """
    return a * b 
```

```python
add_num.invoke({"a":12 , "b":13})
```

Here we learn how we connect the decoratore with llm to perform task 

```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent   # need to help to create agent we use for this 

```

```python
llm = ChatOpenAI(model="gpt-5.4-mini-2026-03-17")
```

```python
agent = create_agent(
    model= llm,
    tools= [add_num , multiply_num],
    system_prompt="You are a math teacher , and always use tools for calculation."
)
```

```python
# response = agent.invoke({"messages": [{"role" : "user" , "content": "What is 2 + 3"}]})
response = agent.invoke({"messages": [{"role" : "user" , "content": "What is 2 + 3 * 3"}]})
```

```python
# response
# response["messages"]
for res in response["messages"]:
    print(res)
```
