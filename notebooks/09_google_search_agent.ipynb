for ai agent making we need two tools llm   and second is search tool or simple we say tool.

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langchain_community.tools import DuckDuckGoSearchRun 
```

```python
model = ChatGroq(model="openai/gpt-oss-20b")
search = DuckDuckGoSearchRun()
```

know we connect llm and search to each other and then build the result 

```python
search.run("Address of Bagh")
```

know we connect with agent so we do like this 

```python
agent = create_agent(
    model = model,
    tools = [search],
    system_prompt="You are a agent and can search for any question on google"
)


# question = " what is the adress of read foundation bagh "
question = " who is the winner of world cup 2025 women's "
response = agent.invoke({"messages":[{"role":"user" , "content":question}]})
# response["messages"]
print(response['messages'][-1].content)
```
