here we know about csv file loader 

problem in csv files and other pandas in loading they use utf enncode but old window directly encode the german names other then english so we need to use in csv loader utf-8-sig for encoding like below

```python
from langchain_community.document_loaders import CSVLoader
loader = CSVLoader("employee_records.csv" , encoding="utf-8-sig")
docs = loader.load()
len(docs)
# docs[:5]
```

Web page loader we can do like this 

we use another package which is beautifulsoup4 pkg help to crwal on html pages ::Beautiful Soup is a library that makes it easy to scrape information from web pages. It sits atop an HTML or XML parser, providing Pythonic idioms for iterating, searching, and modifying the parse tree.

```python
from langchain_community.document_loaders import WebBaseLoader
loader = WebBaseLoader(web_path="https://www.nature.com/articles/d41586-026-03383-7")
docs = loader.load()
# docs[0].metadata
docs[0].page_content
```

WIKIPEDIA loader 

```python
from langchain_community.document_loaders import WikipediaLoader
loader = WikipediaLoader(query="GEN AI " , load_max_docs=2)
docs = loader.load()
docs[0].page_content
 
```
