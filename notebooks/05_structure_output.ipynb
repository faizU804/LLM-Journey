```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-5-mini-2025-08-07")
```

```python
text = "hello my name is faiz Email is faizusardar900@gmail.com and my age is 24"
```

```python
res = llm.invoke(f"please give me only name , email is and age from this text :{text}")
print(res.content)
```

Solution is Structure Output so we do it here with library which is pydantic library we use here to define the base model which should be followed as well.

```python
from pydantic import BaseModel , Field
from typing import List
class ResponseStrecture(BaseModel):
    name:str   = Field(description="Complete name of User")
    email:str  = Field(description="Email of user")
    age:int    = Field(description="Age of User")

strectured_llm = llm.with_structured_output(ResponseStrecture)
```

```python
res = strectured_llm.invoke(f"please give me only name, email and age from this user : {text}")
res.model_dump()
# res.age
# res.name
# res.email
```

Multiple level of structure we can generate just like in production level

```python
class Movies(BaseModel):
    title:str  = Field(description="Movie Title")
    year:int  = Field(description="Movie Year")


class AllMovies(BaseModel):
    movies:list[Movies]


    
movie_llm = llm.with_structured_output(AllMovies)
```

```python
# movie_llm.invoke("Give me one Sci-Fi movie")
res = movie_llm.invoke("Give me funny new movie")
print(res)
```
